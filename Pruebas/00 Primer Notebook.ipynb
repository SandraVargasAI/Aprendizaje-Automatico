# Celda que realiza las Operaciones Matemáticas

```python
a = 5
b = 3
suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b

print("La suma es:", suma)
print("La resta es:", resta)
print("La multiplicacion es:", multiplicacion)
print("La division es:", division)
```

## Importamos biblioteca y usamos función raiz cuadrada

```python
import math

numero = 16
raiz2 = math.sqrt(numero)

print("La raiz de ", numero, "es: ", raiz2)
```

## Uso de FOR en una lista

```python
frutas = ["manzana","banana","naranja","uva"]

print("Lista de frutas:")
for fruta in frutas:
    print(fruta)
```

## Uso de IF

```python
numero = 4

if numero / 2 == 0:
    print(numero, "es par.")
else:
    print(numero, "es impar.")
```

## Gráfico de Barras (Matplotlib)

```python
import matplotlib.pyplot as plt

etiquetas = ['A','B','C','D']
valores = [10, 7, 5, 12]

plt.bar(etiquetas, valores)
plt.xlabel('Categorías')
plt.ylabel('Valores')
plt.title('Gráfico de Barras')
plt.show()
```
